## loading libraries, Datasets and preprocessing  and global variables and functions

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score, roc_curve, precision_score, recall_score, f1_score
import warnings
warnings.filterwarnings('ignore')

# functions (same for all datasets) 
def generate_hmm_sequence(pi, A, mus, sigmas, T, seed=None):
    rng = np.random.RandomState(seed)
    K = len(mus)
    z = np.zeros(T, dtype=int)
    x = np.zeros(T)
    z[0] = rng.choice(K, p=pi)
    x[0] = rng.normal(mus[z[0]], np.sqrt(sigmas[z[0]]))
    for t in range(1, T):
        z[t] = rng.choice(K, p=A[z[t-1]])
        x[t] = rng.normal(mus[z[t]], np.sqrt(sigmas[z[t]]))
    return z, x

def generate_synthetic_dataset(n_nominal, n_anomaly, T, anomaly_block_len, mu_anomaly=4.0, seed=1405):
    rng = np.random.RandomState(seed)
    X_nominal = [rng.normal(0, 1, T) for _ in range(n_nominal)]
    X_anomaly = []
    for _ in range(n_anomaly):
        seq = rng.normal(0, 1, T)
        start = rng.randint(0, T - anomaly_block_len)
        seq[start:start+anomaly_block_len] = rng.normal(mu_anomaly, 1, anomaly_block_len)
        X_anomaly.append(seq)
    return X_nominal, X_anomaly

def log_prob_markov(z, pi, A):
    log_p = np.log(pi[z[0]] + 1e-15)
    for t in range(1, len(z)):
        log_p += np.log(A[z[t-1], z[t]] + 1e-15)
    return log_p

def viterbi_log(x, pi, A, mus, sigmas):
    T_len = len(x)
    K = len(mus)
    log_emission = np.zeros((T_len, K))
    for k in range(K):
        log_emission[:, k] = -0.5 * np.log(2 * np.pi * sigmas[k]) - 0.5 * ((x - mus[k])**2 / sigmas[k])
    delta = np.zeros((T_len, K))
    psi = np.zeros((T_len, K), dtype=int)
    delta[0, :] = np.log(pi + 1e-15) + log_emission[0, :]
    for t in range(1, T_len):
        for j in range(K):
            temp = delta[t-1, :] + np.log(A[:, j] + 1e-15)
            best = np.argmax(temp)
            delta[t, j] = temp[best] + log_emission[t, j]
            psi[t, j] = best
    z_star = np.zeros(T_len, dtype=int)
    z_star[-1] = np.argmax(delta[-1, :])
    for t in range(T_len-2, -1, -1):
        z_star[t] = psi[t+1, z_star[t+1]]
    return z_star

def compute_joint_features(x, z, K):
    T_len = len(x)
    c = np.zeros((K, K))
    for t in range(1, T_len):
        c[z[t-1], z[t]] += 1
    c_flat = c.flatten() / (T_len - 1)
    e = np.zeros(K)
    counts = np.zeros(K)
    for t in range(T_len):
        e[z[t]] += x[t]
        counts[z[t]] += 1
    e = e / (counts + 1e-15)
    return np.concatenate([c_flat, e])

def prepare_features(sequences, mode='flatten', window_size=10):
    X = []
    for seq in sequences:
        if mode == 'flatten':
            X.append(seq)
        elif mode == 'summary':
            X.append([np.mean(seq), np.std(seq), np.min(seq), np.max(seq), np.median(seq)])
        elif mode == 'window':
            feats = []
            for w in range(0, len(seq), window_size):
                window = seq[w:w+window_size]
                feats.append(np.mean(window))
                feats.append(np.std(window))
            X.append(feats)
    return np.array(X)

def evaluate_oc_svm(X_train, X_test, y_test, nu=0.1, gamma='scale', kernel='rbf'):
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    model = OneClassSVM(kernel=kernel, gamma=gamma, nu=nu)
    model.fit(X_train_scaled)
    scores = model.decision_function(X_test_scaled)
    anomaly_scores = -scores
    auc = roc_auc_score(y_test, anomaly_scores)
    return auc, anomaly_scores, model, scaler

def train_hmad(X_train, K=2, nu=0.1, max_iter=10, gamma='scale', seed=1405):
    np.random.seed(seed)
    n_seq = len(X_train)
    T_len = len(X_train[0])
    pooled = np.concatenate(X_train).reshape(-1, 1)
    kmeans = KMeans(n_clusters=K, random_state=seed, n_init=10)
    kmeans.fit(pooled)
    mus = kmeans.cluster_centers_.flatten()
    idx = np.argsort(mus)
    mus = mus[idx]
    sigmas = np.zeros(K)
    labels_km = kmeans.predict(pooled)
    for k in range(K):
        cluster_data = pooled[labels_km == idx[k]]
        sigmas[k] = np.var(cluster_data) if len(cluster_data) > 1 else 1.0
    A = np.full((K, K), 0.1 / (K-1)) if K > 1 else np.array([[1.0]])
    for k in range(K):
        A[k, k] = 0.9
    pi = np.ones(K) / K
    assignment_history = []
    score_history = []
    frac_changed_history = []
    SIGMA_FLOOR = 1e-6
    for m in range(max_iter):
        z_assignments = [viterbi_log(seq, pi, A, mus, sigmas) for seq in X_train]
        features = np.array([compute_joint_features(seq, z, K) for seq, z in zip(X_train, z_assignments)])
        scaler = StandardScaler()
        features_scaled = scaler.fit_transform(features)
        model = OneClassSVM(kernel='rbf', gamma=gamma, nu=nu)
        model.fit(features_scaled)
        scores = model.decision_function(features_scaled)
        score_history.append(np.mean(scores))
        if m > 0:
            changes = sum(not np.array_equal(z_assignments[i], assignment_history[-1][i]) for i in range(n_seq))
            frac_changed = changes / n_seq
        else:
            frac_changed = 1.0
        frac_changed_history.append(frac_changed)
        assignment_history.append(z_assignments)
        if m > 0 and frac_changed < 0.01:
            print(f"HMAD converged at iteration {m+1}")
            break
        A_new = np.zeros((K, K))
        for z in z_assignments:
            for t in range(1, T_len):
                A_new[z[t-1], z[t]] += 1
        for i in range(K):
            row_sum = A_new[i, :].sum()
            A_new[i, :] = A_new[i, :] / row_sum if row_sum > 0 else 1.0/K
        A = A_new
        counts = np.zeros(K)
        sums = np.zeros(K)
        sumsq = np.zeros(K)
        for seq, z in zip(X_train, z_assignments):
            for t in range(T_len):
                k = z[t]
                counts[k] += 1
                sums[k] += seq[t]
                sumsq[k] += seq[t]**2
        for k in range(K):
            if counts[k] > 1:
                mus[k] = sums[k] / counts[k]
                sigmas[k] = max((sumsq[k] / counts[k]) - mus[k]**2, SIGMA_FLOOR)
            else:
                sigmas[k] = 1.0
    return model, scaler, A, mus, sigmas, assignment_history, score_history, frac_changed_history

def predict_hmad(X_test, model, scaler, A, mus, sigmas, K=2):
    features = []
    for seq in X_test:
        z_seq = viterbi_log(seq, np.ones(K)/K, A, mus, sigmas)
        features.append(compute_joint_features(seq, z_seq, K))
    features = np.array(features)
    features_scaled = scaler.transform(features)
    scores = model.decision_function(features_scaled)
    return -scores

def sq7_underflow_demo(T_len=5000, seed=42):
    pi = np.array([0.5, 0.5])
    A = np.array([[0.9, 0.1], [0.1, 0.9]])
    rng = np.random.RandomState(seed)
    z = np.zeros(T_len, dtype=int)
    z[0] = rng.choice(2, p=pi)
    for t in range(1, T_len):
        z[t] = rng.choice(2, p=A[z[t-1]])
    raw_prob = pi[z[0]]
    underflow_t = None
    for t in range(1, T_len):
        raw_prob *= A[z[t-1], z[t]]
        if raw_prob == 0.0 and underflow_t is None:
            underflow_t = t
    log_p = log_prob_markov(z, pi, A)
    print(f"Raw product P(z) after T={T_len} steps: {raw_prob}")
    print(f"Log-space log P(z): {log_p:.4f}")
    if underflow_t is not None:
        print(f"Underflow at t={underflow_t}")
    return raw_prob, log_p, underflow_t

all_results = []
```

```python
# CIC‑IDS2017 
T = 20
TRAIN_SEED = 1405; TEST_SEED = 2026; SQ6_SEED = 7
test_anomaly_fractions = [0.01, 0.05, 0.1, 0.2]
FEATURE_COL = 'Total_Fwd_Packets'

# Data loading
monday    = pd.read_csv('Monday-WorkingHours.pcap_ISCX.csv', skipinitialspace=True)
tuesday   = pd.read_csv('Tuesday-WorkingHours.pcap_ISCX.csv', skipinitialspace=True)
wednesday = pd.read_csv('Wednesday-workingHours.pcap_ISCX.csv', skipinitialspace=True)
for df in [monday, tuesday, wednesday]:
    df.columns = df.columns.str.strip().str.replace(' ', '_')

def prepare_day(df):
    df = df.copy()
    df['binary_label'] = (df['Label'] != 'BENIGN').astype(int)
    if 'Timestamp' in df.columns:
        df['Timestamp'] = pd.to_datetime(df['Timestamp'], errors='coerce')
        df = df.sort_values('Timestamp').reset_index(drop=True)
    return df
monday    = prepare_day(monday)
tuesday   = prepare_day(tuesday)
wednesday = prepare_day(wednesday)

def create_windows_with_mask(df, feature_col):
    values = df[feature_col].values
    labels = df['binary_label'].values
    n = len(values)
    windows, win_labels, attack_masks = [], [], []
    for i in range(0, n - T + 1, T):
        win = values[i:i+T]
        lbl = 1 if np.any(labels[i:i+T] == 1) else 0
        windows.append(win); win_labels.append(lbl)
        attack_masks.append(labels[i:i+T] == 1)
    return np.array(windows), np.array(win_labels), attack_masks

X_mon, y_mon, mask_mon = create_windows_with_mask(monday, FEATURE_COL)
X_tue, y_tue, mask_tue = create_windows_with_mask(tuesday, FEATURE_COL)
X_wed, y_wed, mask_wed = create_windows_with_mask(wednesday, FEATURE_COL)

np.random.seed(TRAIN_SEED)
n_train = min(200, len(X_mon))
train_idx = np.random.choice(len(X_mon), size=n_train, replace=False)
X_train = [X_mon[i] for i in train_idx]

mask_mon_pool = np.ones(len(X_mon), dtype=bool); mask_mon_pool[train_idx] = False
X_test_pool = [X_mon[i] for i in np.where(mask_mon_pool)[0]]
y_test_pool = [0] * len(X_test_pool)
test_masks = [mask_mon[i] for i in range(len(mask_mon)) if mask_mon_pool[i]]

attack_idx_tue = np.where(y_tue == 1)[0]
X_test_pool.extend([X_tue[i] for i in attack_idx_tue])
y_test_pool.extend([1] * len(attack_idx_tue))
test_masks.extend([mask_tue[i] for i in attack_idx_tue])

attack_idx_wed = np.where(y_wed == 1)[0]
X_test_pool.extend([X_wed[i] for i in attack_idx_wed])
y_test_pool.extend([1] * len(attack_idx_wed))
test_masks.extend([mask_wed[i] for i in attack_idx_wed])

X_test_pool = np.array(X_test_pool)
y_test_pool = np.array(y_test_pool)
normal_test_idx = np.where(y_test_pool == 0)[0]
anom_test_idx   = np.where(y_test_pool == 1)[0]

# HMAD training 
print("Training HMAD for CIC‑IDS2017 ")
hmad_model, scaler_hmad, A_final, mus_final, sigmas_final, _, _, _ = train_hmad(
    X_train, K=2, nu=0.1, max_iter=22, seed=TRAIN_SEED)

# Multi‑baseline
window_size = max(1, T // 5)
X_train_flat = prepare_features(X_train, 'flatten')
X_train_sum  = prepare_features(X_train, 'summary')
X_train_win  = prepare_features(X_train, 'window', window_size)

results_flat, results_sum, results_win, results_hmad = [], [], [], []
roc_data = {}

for r_test in test_anomaly_fractions:
    n_anom = int(100 * r_test)
    n_norm = 100 - n_anom
    n_anom = min(n_anom, len(anom_test_idx))
    n_norm = min(n_norm, len(normal_test_idx))
    np.random.seed(TEST_SEED)
    ch_n = np.random.choice(normal_test_idx, size=n_norm, replace=False)
    ch_a = np.random.choice(anom_test_idx, size=n_anom, replace=False) if n_anom > 0 else []
    X_cur = list(X_test_pool[ch_n]) + list(X_test_pool[ch_a])
    y_cur = [0]*n_norm + [1]*n_anom

    # OC‑SVM baselines
    auc_f, scores_f, _, _ = evaluate_oc_svm(X_train_flat, prepare_features(X_cur, 'flatten'), y_cur, nu=0.1)
    auc_s, scores_s, _, _ = evaluate_oc_svm(X_train_sum, prepare_features(X_cur, 'summary'), y_cur, nu=0.1)
    auc_w, scores_w, _, _ = evaluate_oc_svm(X_train_win, prepare_features(X_cur, 'window', window_size), y_cur, nu=0.1)
    results_flat.append(auc_f); results_sum.append(auc_s); results_win.append(auc_w)

    # HMAD
    scores_h = predict_hmad(X_cur, hmad_model, scaler_hmad, A_final, mus_final, sigmas_final, K=2)
    auc_h = roc_auc_score(y_cur, scores_h)
    results_hmad.append(auc_h)

    print(f"r_test={r_test:.2f}: Flat={auc_f:.4f}  Sum={auc_s:.4f}  Win={auc_w:.4f}  HMAD={auc_h:.4f}")
    if abs(r_test - 0.1) < 1e-6:
        roc_data = {'y': y_cur, 'flat': scores_f, 'sum': scores_s, 'win': scores_w, 'hmad': scores_h,
                     'X_test': X_cur, 'local_anom_idx': n_norm, 'global_anom_idx': ch_a}


if roc_data:
    plt.figure(figsize=(8,6))
    fpr_f, tpr_f, _ = roc_curve(roc_data['y'], roc_data['flat'])
    fpr_s, tpr_s, _ = roc_curve(roc_data['y'], roc_data['sum'])
    fpr_w, tpr_w, _ = roc_curve(roc_data['y'], roc_data['win'])
    fpr_h, tpr_h, _ = roc_curve(roc_data['y'], roc_data['hmad'])
    plt.plot(fpr_f, tpr_f, '--', label=f'OC-SVM Flattened (AUC={roc_auc_score(roc_data["y"], roc_data["flat"]):.3f})')
    plt.plot(fpr_s, tpr_s, ':', label=f'OC-SVM Summary (AUC={roc_auc_score(roc_data["y"], roc_data["sum"]):.3f})')
    plt.plot(fpr_w, tpr_w, '-.', label=f'OC-SVM Window (AUC={roc_auc_score(roc_data["y"], roc_data["win"]):.3f})')
    plt.plot(fpr_h, tpr_h, label=f'HMAD (AUC={roc_auc_score(roc_data["y"], roc_data["hmad"]):.3f})')
    plt.plot([0,1],[0,1],'k--'); plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
    plt.title('ROC Curves – CIC‑IDS2017 (r_test=0.1)')
    plt.legend(); plt.grid(alpha=0.3); plt.show()

    # Sequence visualisation with attack shade
    y_true, pred_hmad = np.array(roc_data['y']), (roc_data['hmad'] > 0).astype(int)
    norm_idx = np.where((y_true == 0) & (pred_hmad == 0))[0][0]
    anom_local = roc_data['local_anom_idx']
    if not (y_true[anom_local] == 1 and pred_hmad[anom_local] == 1):
        correct_anom = np.where((y_true == 1) & (pred_hmad == 1))[0]
        anom_local = correct_anom[0] if len(correct_anom) > 0 else anom_local
    seq_norm = X_test_pool[norm_idx]
    seq_anom = roc_data['X_test'][anom_local]

    global_idx = roc_data['global_anom_idx'][anom_local - roc_data['local_anom_idx']] if anom_local >= roc_data['local_anom_idx'] else anom_test_idx[0]
    mask_anom = test_masks[global_idx]

    fig, axes = plt.subplots(1, 2, figsize=(14,4))
    axes[0].plot(seq_norm, 'b-'); axes[0].set_title('Normal Sequence (CIC‑IDS2017)'); axes[0].grid(alpha=0.3)
    axes[1].plot(seq_anom, 'b-')
    in_region = False
    for i in range(T):
        if mask_anom[i] and not in_region: start = i; in_region = True
        elif not mask_anom[i] and in_region: axes[1].axvspan(start, i, color='red', alpha=0.3); in_region = False
    if in_region: axes[1].axvspan(start, T, color='red', alpha=0.3)
    axes[1].set_title('Anomalous Sequence (attack flows shaded)'); axes[1].grid(alpha=0.3)
    fig.suptitle('CIC‑IDS2017 – Sequence Visualization (r_test=0.1)', fontsize=14)
    plt.tight_layout(); plt.show()

    # Viterbi state paths
    pi = np.ones(2)/2
    z_norm = viterbi_log(seq_norm, pi, A_final, mus_final, sigmas_final)
    z_anom = viterbi_log(seq_anom, pi, A_final, mus_final, sigmas_final)
    fig2, ax2 = plt.subplots(2,2,figsize=(12,5),sharex='col')
    ax2[0,0].plot(seq_norm,'b-'); ax2[0,0].set_title('Normal')
    ax2[1,0].step(range(len(seq_norm)), z_norm, 'r-', where='mid')
    ax2[0,1].plot(seq_anom,'b-'); ax2[0,1].set_title('Anomalous')
    in_region=False
    for i in range(T):
        if mask_anom[i] and not in_region: start=i; in_region=True
        elif not mask_anom[i] and in_region: ax2[0,1].axvspan(start,i,color='red',alpha=0.3); in_region=False
    if in_region: ax2[0,1].axvspan(start,T,color='red',alpha=0.3)
    ax2[1,1].step(range(len(seq_anom)), z_anom, 'r-', where='mid')
    ax2[1,0].set_ylabel('State'); ax2[1,1].set_ylabel('State')
    plt.suptitle('HMAD State Paths – CIC‑IDS2017', fontsize=14)
    plt.tight_layout(); plt.show()

# Contamination sensitivity
plt.figure(figsize=(8,5))
plt.plot(test_anomaly_fractions, results_flat, 'o--', label='OC-SVM Flattened')
plt.plot(test_anomaly_fractions, results_sum, 's:', label='OC-SVM Summary')
plt.plot(test_anomaly_fractions, results_win, 'D-.', label='OC-SVM Window')
plt.plot(test_anomaly_fractions, results_hmad, '^-', label='HMAD')
plt.xlabel('Test anomaly fraction'); plt.ylabel('ROC-AUC')
plt.title('CIC‑IDS2017 – Contamination Sensitivity')
plt.legend(); plt.grid(alpha=0.4); plt.show()

all_results.append({
    'dataset': 'CIC-IDS2017',
    'OC_AUCs': results_flat,
    'HMAD_AUCs': results_hmad,
    'fracs': test_anomaly_fractions
})
```

```python
# DNA 
T = 20
TRAIN_SEED = 1405; TEST_SEED = 2026; SQ6_SEED = 7
test_anomaly_fractions = [0.01, 0.05, 0.1, 0.2]

df_dna = pd.read_csv('datasetspart1/train2.csv')
dna_string = ''.join(df_dna['sequence'].values)
mapping = {'A':0, 'C':1, 'G':2, 'T':3}
int_seq = np.array([mapping[base] for base in dna_string if base in mapping])

# Create windows
X_all = []
for i in range(0, len(int_seq) - T + 1, T):
    X_all.append(int_seq[i:i+T])
X_all = np.array(X_all)

np.random.seed(TRAIN_SEED)
n_train = min(200, len(X_all))
train_idx = np.random.choice(len(X_all), size=n_train, replace=False)
X_train = [X_all[i] for i in train_idx]

def mutate_window(win, mutation_rate=0.05):
    mutated = win.copy()
    n_mut = max(1, int(len(win) * mutation_rate))
    pos = np.random.choice(len(win), size=n_mut, replace=False)
    for p in pos:
        mutated[p] = (mutated[p] + np.random.randint(1,4)) % 4
    mask = np.zeros(len(win), dtype=bool)
    mask[pos] = True
    return mutated, mask

# Test pool
mask_all = np.ones(len(X_all), dtype=bool); mask_all[train_idx] = False
X_normal_test = X_all[mask_all]
n_anom = min(100, len(X_normal_test))
np.random.seed(TEST_SEED)
anom_idx = np.random.choice(len(X_normal_test), size=n_anom, replace=False)

X_anom_test = []
masks_anom_test = []
for i in anom_idx:
    mut_seq, mut_mask = mutate_window(X_normal_test[i])
    X_anom_test.append(mut_seq)
    masks_anom_test.append(mut_mask)
X_anom_test = np.array(X_anom_test)
masks_anom_test = np.array(masks_anom_test)
X_normal_test = np.delete(X_normal_test, anom_idx, axis=0)

X_test_pool = np.concatenate([X_normal_test, X_anom_test])
y_test_pool = np.array([0]*len(X_normal_test) + [1]*len(X_anom_test))
test_masks = [np.zeros(T, dtype=bool)]*len(X_normal_test) + [masks_anom_test[i] for i in range(len(X_anom_test))]
normal_test_idx = np.where(y_test_pool == 0)[0]
anom_test_idx   = np.where(y_test_pool == 1)[0]

print(f"Train normal windows: {len(X_train)}")
print(f"Test pool: {len(X_test_pool)} total, {y_test_pool.sum()} anomalies (frac {y_test_pool.mean():.3f})")

# HMAD training
print("Training HMAD for DNA")
hmad_model, scaler_hmad, A_final, mus_final, sigmas_final, _, _, _ = train_hmad(
    X_train, K=2, nu=0.1, max_iter=22, seed=TRAIN_SEED)

# Baselines
window_size = max(1, T//5)
X_train_flat = prepare_features(X_train, 'flatten')
X_train_sum  = prepare_features(X_train, 'summary')
X_train_win  = prepare_features(X_train, 'window', window_size)

results_flat, results_sum, results_win, results_hmad = [], [], [], []
roc_data = {}

for r_test in test_anomaly_fractions:
    n_anom = int(100 * r_test)
    n_norm = 100 - n_anom
    n_anom = min(n_anom, len(anom_test_idx))
    n_norm = min(n_norm, len(normal_test_idx))
    np.random.seed(TEST_SEED)
    ch_n = np.random.choice(normal_test_idx, n_norm, replace=False)
    ch_a = np.random.choice(anom_test_idx, n_anom, replace=False) if n_anom>0 else []
    X_cur = list(X_test_pool[ch_n]) + list(X_test_pool[ch_a])
    y_cur = [0]*n_norm + [1]*n_anom

    # OC‑SVM
    auc_f, scores_f, _, _ = evaluate_oc_svm(X_train_flat, prepare_features(X_cur, 'flatten'), y_cur, nu=0.1)
    auc_s, scores_s, _, _ = evaluate_oc_svm(X_train_sum, prepare_features(X_cur, 'summary'), y_cur, nu=0.1)
    auc_w, scores_w, _, _ = evaluate_oc_svm(X_train_win, prepare_features(X_cur, 'window', window_size), y_cur, nu=0.1)
    results_flat.append(auc_f); results_sum.append(auc_s); results_win.append(auc_w)

    # HMAD
    scores_h = predict_hmad(X_cur, hmad_model, scaler_hmad, A_final, mus_final, sigmas_final, K=2)
    auc_h = roc_auc_score(y_cur, scores_h)
    results_hmad.append(auc_h)

    print(f"r_test={r_test:.2f}: Flat={auc_f:.4f}  Sum={auc_s:.4f}  Win={auc_w:.4f}  HMAD={auc_h:.4f}")
    if abs(r_test - 0.1) < 1e-6:
        roc_data = {'y': y_cur, 'flat': scores_f, 'sum': scores_s, 'win': scores_w, 'hmad': scores_h,
                     'X_test': X_cur, 'local_anom_idx': n_norm, 'global_anom_idx': ch_a}

if roc_data:
    plt.figure(figsize=(8,6))
    fpr_f, tpr_f, _ = roc_curve(roc_data['y'], roc_data['flat'])
    fpr_s, tpr_s, _ = roc_curve(roc_data['y'], roc_data['sum'])
    fpr_w, tpr_w, _ = roc_curve(roc_data['y'], roc_data['win'])
    fpr_h, tpr_h, _ = roc_curve(roc_data['y'], roc_data['hmad'])
    plt.plot(fpr_f, tpr_f, '--', label=f'OC-SVM Flat (AUC={roc_auc_score(roc_data["y"], roc_data["flat"]):.3f})')
    plt.plot(fpr_s, tpr_s, ':', label=f'OC-SVM Sum (AUC={roc_auc_score(roc_data["y"], roc_data["sum"]):.3f})')
    plt.plot(fpr_w, tpr_w, '-.', label=f'OC-SVM Win (AUC={roc_auc_score(roc_data["y"], roc_data["win"]):.3f})')
    plt.plot(fpr_h, tpr_h, label=f'HMAD (AUC={roc_auc_score(roc_data["y"], roc_data["hmad"]):.3f})')
    plt.plot([0,1],[0,1],'k--')
    plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
    plt.title('ROC Curves – DNA (r_test=0.1)')
    plt.legend(); plt.grid(alpha=0.3); plt.show()

    y_true = np.array(roc_data['y'])
    pred_hmad = (roc_data['hmad'] > 0).astype(int)
    norm_idx = np.where((y_true == 0) & (pred_hmad == 0))[0][0]
    anom_local = roc_data['local_anom_idx']
    if not (y_true[anom_local] == 1 and pred_hmad[anom_local] == 1):
        correct_anom = np.where((y_true == 1) & (pred_hmad == 1))[0]
        anom_local = correct_anom[0] if len(correct_anom) > 0 else anom_local
    seq_norm = X_test_pool[norm_idx]
    seq_anom = roc_data['X_test'][anom_local]

    global_idx = roc_data['global_anom_idx'][anom_local - roc_data['local_anom_idx']] if anom_local >= roc_data['local_anom_idx'] else anom_test_idx[0]
    mask_anom = test_masks[global_idx]
    change_points = np.where(np.diff(np.concatenate(([0], mask_anom.astype(int), [0]))))[0]

    fig_a, axes_a = plt.subplots(1, 2, figsize=(14,4))
    axes_a[0].plot(seq_norm, color='navy')
    axes_a[0].set_title('Nominal DNA Sequence (Encoded)')
    axes_a[0].set_ylabel('Nucleotide Code')
    axes_a[0].grid(alpha=0.3)

    axes_a[1].plot(seq_anom, color='navy')
    for i in range(0, len(change_points), 2):
        axes_a[1].axvspan(change_points[i], change_points[i+1], alpha=0.25, color='#FF6B6B',
                          label='True Mutation Block' if i==0 else "")
    axes_a[1].set_title('Anomalous DNA Sequence (Mutations Highlighted)')
    axes_a[1].set_ylabel('Nucleotide Code')
    axes_a[1].legend(loc='upper right')
    axes_a[1].grid(alpha=0.3)
    fig_a.suptitle('a) DNA Sequence Visualization (r_test=0.1)', fontsize=14)
    plt.tight_layout(); plt.show()

    # Viterbi state paths
    pi = np.ones(2)/2
    z_norm = viterbi_log(seq_norm, pi, A_final, mus_final, sigmas_final)
    z_anom = viterbi_log(seq_anom, pi, A_final, mus_final, sigmas_final)

    fig_d, axes_d = plt.subplots(2, 2, figsize=(12,6), sharex='col')
    axes_d[0,0].plot(seq_norm, color='navy'); axes_d[0,0].set_title('Normal')
    axes_d[1,0].step(np.arange(len(seq_norm)), z_norm, where="mid", color='crimson', lw=2)
    axes_d[1,0].set_ylim(-0.2, 1.2); axes_d[1,0].set_ylabel('Viterbi State')

    axes_d[0,1].plot(seq_anom, color='navy')
    for i in range(0, len(change_points), 2):
        axes_d[0,1].axvspan(change_points[i], change_points[i+1], alpha=0.25, color='#FF6B6B')
    axes_d[0,1].set_title('Anomalous')
    axes_d[1,1].step(np.arange(len(seq_anom)), z_anom, where="mid", color='crimson', lw=2)
    axes_d[1,1].set_ylim(-0.2, 1.2); axes_d[1,1].set_ylabel('Viterbi State')
    plt.suptitle(f'd) HMAD State Transitions (K=2) on DNA Mutations', fontsize=14)
    plt.tight_layout(); plt.show()

# Contamination sensitivity
plt.figure(figsize=(8,5))
plt.plot(test_anomaly_fractions, results_flat, 'o--', label='OC-SVM Flattened')
plt.plot(test_anomaly_fractions, results_sum, 's:', label='OC-SVM Summary')
plt.plot(test_anomaly_fractions, results_win, 'D-.', label='OC-SVM Window')
plt.plot(test_anomaly_fractions, results_hmad, '^-', label='HMAD')
plt.xlabel('Test Anomaly Fraction'); plt.ylabel('ROC-AUC')
plt.title('DNA – Contamination Sensitivity')
plt.legend(); plt.grid(alpha=0.4); plt.show()

all_results.append({
    'dataset': 'DNA',
    'OC_AUCs': results_flat,
    'HMAD_AUCs': results_hmad,
    'fracs': test_anomaly_fractions
})
```

## Theoretical Question 10. 
Explain why a detector that only looks at individual observations may fail to detect a sequence anomaly. Give one example where each individual observation is normal, but the ordering is suspicious.



آشکارسازی که مشاهدات را مستقل فرض کند تنها بررسی می‌کند که آیا هر مقدار به‌تنهایی تحت توزیع نرمال محتمل است یا خیر. این آشکارساز ترتیب و گذارهای بین مشاهدات را نادیده می‌گیرد.درحالی که یک ناهنجاری توالی ممکن است کاملاً از مقادیر منفرد عادی تشکیل شده باشد که در یک الگوی زمانی غیرعادی چیده شده‌اند؛ چنین ناهنجاری‌ای از دید این آشکارساز پنهان می‌ماند.

به عنوان مثال: 

ضربان قلب یک بیمار (بر حسب تپش در دقیقه) که هر دقیقه اندازه‌گیری می‌شود:

\[
۷۲ $\to$ ۷۳ $\to$ ۷۵ $\to$ ۷۷ $\to$ ۸۰ $\to$ ۸۵ $\to$ ۹۲ $\to$ ۱۰۰ $\to$ ۱۰۸ $\to$ ۱۱۵ $\to$ ۱۱۰ $\to$ ۸۰ $\to$ ۷۴
\]

تک‌تک این مقادیر (۷۲، ۷۳، …، ۱۱۵) در محدوده‌ی محتمل برای یک فرد در حال فعالیت ملایم یا استراحت قرار دارند. آشکارسازی که هر مقدار را مستقل ببیند، هیچ مورد غیرعادی‌ای پیدا نمی‌کند. اما روند افزایشی پیوسته در طی ۱۰ دقیقه و سپس افت ناگهانی از نظر فیزیولوژیکی مشکوک است – ممکن است نشانه‌ی یک آریتمی، اختلال سیستم عصبی خودکار، یا خطای حسگر باشد. و درواقع ناهنجاری در الگوی روند و گذارها نهفته است، نه در اعداد منفرد.

## Theoretical Question 17. 
A stationary distribution v of a Markov chain satisfies v⊤A =v⊤,
For the transition matrix sigma vi = 1.
A= [0.95 0.05
0.10 0.90 ]

compute the stationary distribution and interpret it as the long-run fraction of time spent in each
state.

ماتریس انتقال 

A = \begin{pmatrix} 0.95 & 0.05 \\ 0.10 & 0.90 \end{pmatrix}

و توزیع پایدار $\mathbf{v} = (v_1, v_2)^\top$ در شرط زیر صدق می‌کند:


$
\mathbf{v}^\top A = \mathbf{v}^\top, \qquad v_1 + v_2 = 1.
$

بدین ترتیب با نوشتن دو مؤلفه‌ی معادله‌ی $\mathbf{v}^\top A = \mathbf{v}^\top$ داریم:

$
\begin{aligned}
v_1 &= 0.95\,v_1 + 0.10\,v_2,\\
v_2 &= 0.05\,v_1 + 0.90\,v_2.
\end{aligned}
$

از معادله‌ی اول:

$
v_1 - 0.95\,v_1 = 0.10\,v_2 \quad\Longrightarrow\quad 0.05\,v_1 = 0.10\,v_2 \quad\Longrightarrow\quad v_1 = 2\,v_2.
$

با جایگذاری در شرط مجموع:

$
v_1 + v_2 = 2\,v_2 + v_2 = 3\,v_2 = 1 \quad\Longrightarrow\quad v_2 = \frac{1}{3} \approx 0.333.
$

در نتیجه:

$
v_1 = 2\,v_2 = \frac{2}{3} \approx 0.667.
$

 بدین ترتیب در بلند مدت زنجیره‌ی مارکوف تقریباً دو‌سوم زمان را در وضعیت ۱ و یک‌سوم زمان را در وضعیت ۲ سپری می‌کند
