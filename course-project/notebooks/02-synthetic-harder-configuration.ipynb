## loading libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score, roc_curve, precision_score, recall_score, f1_score
import warnings
warnings.filterwarnings('ignore')
```

## functions

```python
# OC-SVM BASELINE (SQ1, SQ2, SQ3)
def prepare_features(sequences, mode='flatten', window_size=10):
    X = []
    for seq in sequences:
        if mode == 'flatten':
            X.append(seq)
        elif mode == 'summary':
            X.append([np.mean(seq), np.std(seq), np.min(seq), np.max(seq), np.median(seq)])
        elif mode == 'window':
            feats = []
            for w in range(0, len(seq), window_size):
                window = seq[w:w+window_size]
                feats.append(np.mean(window))
                feats.append(np.std(window))
            X.append(feats)
    return np.array(X)

def evaluate_oc_svm(X_train, X_test, y_test, nu=0.1, gamma='scale', kernel='rbf'):
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    model = OneClassSVM(kernel=kernel, gamma=gamma, nu=nu)
    model.fit(X_train_scaled)
    scores = model.decision_function(X_test_scaled)  
    anomaly_scores = -scores  
    auc = roc_auc_score(y_test, anomaly_scores)
    return auc, anomaly_scores, model, scaler




# DATA GENERATION (SQ5) 

def generate_hmm_sequence(pi, A, mus, sigmas, T, seed=None):
    rng = np.random.RandomState(seed)
    K = len(mus)
    z = np.zeros(T, dtype=int)
    x = np.zeros(T)
    z[0] = rng.choice(K, p=pi)
    x[0] = rng.normal(mus[z[0]], np.sqrt(sigmas[z[0]]))
    for t in range(1, T):
        z[t] = rng.choice(K, p=A[z[t-1]])
        x[t] = rng.normal(mus[z[t]], np.sqrt(sigmas[z[t]]))
    return z, x

def generate_synthetic_dataset(n_nominal, n_anomaly, T, anomaly_block_len, mu_anomaly=4.0, seed=1405):
    rng = np.random.RandomState(seed)
    X_nominal = [rng.normal(0, 1, T) for _ in range(n_nominal)]
    X_anomaly = []
    for _ in range(n_anomaly):
        seq = rng.normal(0, 1, T)
        start = rng.randint(0, T - anomaly_block_len)
        seq[start:start+anomaly_block_len] = rng.normal(mu_anomaly, 1, anomaly_block_len)
        X_anomaly.append(seq)
    return X_nominal, X_anomaly


# HMM TOOLS (SQ4, SQ6)
def log_prob_markov(z, pi, A):
    log_p = np.log(pi[z[0]] + 1e-15)
    for t in range(1, len(z)):
        log_p += np.log(A[z[t-1], z[t]] + 1e-15)
    return log_p

def viterbi_log(x, pi, A, mus, sigmas):
    T = len(x)
    K = len(mus)
    log_emission = np.zeros((T, K))
    for k in range(K):
        log_emission[:, k] = -0.5 * np.log(2 * np.pi * sigmas[k]) - 0.5 * ((x - mus[k])**2 / sigmas[k])

    delta = np.zeros((T, K))
    psi = np.zeros((T, K), dtype=int)
    delta[0, :] = np.log(pi + 1e-15) + log_emission[0, :]
    for t in range(1, T):
        for j in range(K):
            temp = delta[t-1, :] + np.log(A[:, j] + 1e-15)
            best = np.argmax(temp)
            delta[t, j] = temp[best] + log_emission[t, j]
            psi[t, j] = best
    z_star = np.zeros(T, dtype=int)
    z_star[-1] = np.argmax(delta[-1, :])
    for t in range(T-2, -1, -1):
        z_star[t] = psi[t+1, z_star[t+1]]
    return z_star


# raw product vs log-space (SQ7)
def sq7_underflow_demo(T=5000, seed=42):

    pi = np.array([0.5, 0.5])
    A = np.array([[0.9, 0.1],
                   [0.1, 0.9]])
    rng = np.random.RandomState(seed)
    z = np.zeros(T, dtype=int)
    z[0] = rng.choice(2, p=pi)
    for t in range(1, T):
        z[t] = rng.choice(2, p=A[z[t-1]])

    raw_prob = pi[z[0]]
    underflow_t = None
    for t in range(1, T):
        raw_prob *= A[z[t-1], z[t]]
        if raw_prob == 0.0 and underflow_t is None:
            underflow_t = t

    log_p = log_prob_markov(z, pi, A)

    print(f"Raw product P(z) after T={T} steps: {raw_prob}")
    print(f"Log-space log P(z): {log_p:.4f}  (equivalent P(z) ~ {np.exp(log_p) if log_p > -700 else 0.0})")
    if underflow_t is not None:
        print(f"Raw product underflowed to exactly 0.0 at time step t={underflow_t} "
              f"(out of T={T}); log-space computation remained valid throughout.")
    else:
        print("Raw product did not underflow to exactly 0 for this T (try a larger T).")
    return raw_prob, log_p, underflow_t
    

# JOINT FEATURE MAP (SQ9)
def compute_joint_features(x, z, K):
    T = len(x)
    c = np.zeros((K, K))
    for t in range(1, T):
        c[z[t-1], z[t]] += 1
    c_flat = c.flatten() / (T - 1)  
    e = np.zeros(K)
    counts = np.zeros(K)
    for t in range(T):
        e[z[t]] += x[t]
        counts[z[t]] += 1
    e = e / (counts + 1e-15)

    return np.concatenate([c_flat, e])



# HMAD TRAINING (SQ10)
def train_hmad(X_train, K=2, nu=0.1, max_iter=10, gamma='scale', seed=1405):
    np.random.seed(seed)
    n_seq = len(X_train)
    T = len(X_train[0])

    pooled = np.concatenate(X_train).reshape(-1, 1)
    kmeans = KMeans(n_clusters=K, random_state=seed, n_init=10)
    kmeans.fit(pooled)
    mus = kmeans.cluster_centers_.flatten()
    
    idx = np.argsort(mus)
    mus = mus[idx]
    
    sigmas = np.zeros(K)
    labels = kmeans.predict(pooled)
    for k in range(K):
        cluster_data = pooled[labels == idx[k]]
        sigmas[k] = np.var(cluster_data) if len(cluster_data) > 1 else 1.0

    A = np.full((K, K), 0.1 / (K-1)) if K > 1 else np.array([[1.0]])
    for k in range(K):
        A[k, k] = 0.9

    pi = np.ones(K) / K

    assignment_history = []   
    score_history = []
    frac_changed_history = [] 

    SIGMA_FLOOR = 1e-6

    for m in range(max_iter):
        z_assignments = []
        for seq in X_train:
            z_seq = viterbi_log(seq, pi, A, mus, sigmas)
            z_assignments.append(z_seq)

        features = []
        for seq, z_seq in zip(X_train, z_assignments):
            features.append(compute_joint_features(seq, z_seq, K))
        features = np.array(features)

        scaler = StandardScaler()
        features_scaled = scaler.fit_transform(features)
        model = OneClassSVM(kernel='rbf', gamma=gamma, nu=nu)
        model.fit(features_scaled)

        scores = model.decision_function(features_scaled)
        score_history.append(np.mean(scores))

        if m > 0:
            changes = 0
            for i in range(n_seq):
                if not np.array_equal(z_assignments[i], assignment_history[-1][i]):
                    changes += 1
            frac_changed = changes / n_seq
        else:
            frac_changed = 1.0  
        frac_changed_history.append(frac_changed)
        assignment_history.append(z_assignments)

        if m > 0 and frac_changed < 0.01:
            print(f"HMAD converged at iteration {m+1}")
            break
        
        A_new = np.zeros((K, K))
        for z_seq in z_assignments:
            for t in range(1, len(z_seq)):
                A_new[z_seq[t-1], z_seq[t]] += 1
        for i in range(K):
            row_sum = np.sum(A_new[i, :])
            if row_sum > 0:
                A_new[i, :] /= row_sum
            else:
                A_new[i, :] = 1.0 / K
        A = A_new

        counts = np.zeros(K)
        sums = np.zeros(K)
        sumsq = np.zeros(K)
        for seq, z_seq in zip(X_train, z_assignments):
            for t in range(len(seq)):
                k = z_seq[t]
                counts[k] += 1
                sums[k] += seq[t]
                sumsq[k] += seq[t]**2
        for k in range(K):
            if counts[k] > 1:
                mus[k] = sums[k] / counts[k]
                sigmas[k] = max((sumsq[k] / counts[k]) - mus[k]**2, SIGMA_FLOOR)  
            else:
                sigmas[k] = 1.0  

    return model, scaler, A, mus, sigmas, assignment_history, score_history, frac_changed_history


# HMAD PREDICTION (SQ11) 
def predict_hmad(X_test, model, scaler, A, mus, sigmas, K=2):
    
    features = []
    for seq in X_test:
        z_seq = viterbi_log(seq, np.ones(K)/K, A, mus, sigmas)
        feat = compute_joint_features(seq, z_seq, K)
        features.append(feat)
    features = np.array(features)
    features_scaled = scaler.transform(features)
    scores = model.decision_function(features_scaled)  
    return -scores  
```

```python
# SQ4: Log-probability of a Markov state sequence
def log_prob_markov(z, pi, A):
    log_p = np.log(pi[z[0]] + 1e-15)
    for t in range(1, len(z)):
        log_p += np.log(A[z[t-1], z[t]] + 1e-15)
    return log_p

z_test = np.array([0, 1, 0, 1])
pi_test_markov = np.array([0.5, 0.5])
A_test_markov = np.array([[0.9, 0.1], [0.1, 0.9]])
print(f"Markov log-prob (SQ4) test: {log_prob_markov(z_test, pi_test_markov, A_test_markov):.4f}")
```

## Simulation Question 5. 
Implement a function that generates a sequence from a two-state Gaussian-emission HMM. Your function should take π, A, emission means (µ1,µ2), emission variances (σ1^2,σ2^2), and sequence length T. It should return both the hidden state sequence z and the observed sequence x.

```python
# SQ5: Generate a sequence from a 2-state Gaussian HMM
def generate_hmm_sequence(pi, A, mus, sigmas, T, seed=None):
    rng = np.random.RandomState(seed)
    K = len(mus)
    z = np.zeros(T, dtype=int)
    x = np.zeros(T)
    # Initial state
    z[0] = rng.choice(K, p=pi)
    x[0] = rng.normal(mus[z[0]], np.sqrt(sigmas[z[0]]))
    for t in range(1, T):
        z[t] = rng.choice(K, p=A[z[t-1]])
        x[t] = rng.normal(mus[z[t]], np.sqrt(sigmas[z[t]]))
    return z, x
```

## Simulation Question 6. 
Implementlog-space Viterbi decoding for a two-state Gaussian-emission HMM. Test it on a sequence generated by your HMM generator. Report the fraction of time steps where the decoded hidden state matches the true generated hidden state.

```python
# SQ6: Log-space Viterbi decoder
def viterbi_log(x, pi, A, mus, sigmas):
    T = len(x)
    K = len(mus)
    log_emission = np.zeros((T, K))
    for k in range(K):
        log_emission[:, k] = -0.5 * np.log(2 * np.pi * sigmas[k]) - 0.5 * ((x - mus[k])**2 / sigmas[k])

    delta = np.zeros((T, K))
    psi = np.zeros((T, K), dtype=int)
    delta[0, :] = np.log(pi + 1e-15) + log_emission[0, :]
    for t in range(1, T):
        for j in range(K):
            temp = delta[t-1, :] + np.log(A[:, j] + 1e-15)
            best = np.argmax(temp)
            delta[t, j] = temp[best] + log_emission[t, j]
            psi[t, j] = best
    z_star = np.zeros(T, dtype=int)
    z_star[-1] = np.argmax(delta[-1, :])
    for t in range(T-2, -1, -1):
        z_star[t] = psi[t+1, z_star[t+1]]
    return z_star

# Accuracy test (SQ6)
print("\n SQ6: Viterbi Decoding Accuracy")
pi_test = np.array([0.5, 0.5])
A_test = np.array([[0.9, 0.1],
                   [0.1, 0.9]])
mus_test = np.array([0.0, 4.0])
sigmas_test = np.array([1.0, 1.0])
z_true, x_test = generate_hmm_sequence(pi_test, A_test, mus_test, sigmas_test, T=100, seed=SQ6_SEED)
z_pred = viterbi_log(x_test, pi_test, A_test, mus_test, sigmas_test)
acc = np.mean(z_pred == z_true)
print(f"Viterbi accuracy (matching true states): {acc:.2%}")
```

## Simulation Question 7. 
Compare ordinary probability multiplication and log-space computation for a long Markov chain sequence. Generate a sequence of length T = 5000, compute its probability both as a raw product and as a log-probability sum, and show when the raw product begins to suffer from numerical underflow.

```python
# SQ7: Compare raw product and log-space computation
def sq7_underflow_demo(T=5000, seed=42):
    pi = np.array([0.5, 0.5])
    A = np.array([[0.9, 0.1],
                   [0.1, 0.9]])
    rng = np.random.RandomState(seed)
    z = np.zeros(T, dtype=int)
    z[0] = rng.choice(2, p=pi)
    for t in range(1, T):
        z[t] = rng.choice(2, p=A[z[t-1]])

    raw_prob = pi[z[0]]
    underflow_t = None
    for t in range(1, T):
        raw_prob *= A[z[t-1], z[t]]
        if raw_prob == 0.0 and underflow_t is None:
            underflow_t = t

    log_p = log_prob_markov(z, pi, A)

    print(f"Raw product P(z) after T={T} steps: {raw_prob}")
    print(f"Log-space log P(z): {log_p:.4f}  (equivalent P(z) ~ {np.exp(log_p) if log_p > -700 else 0.0})")
    if underflow_t is not None:
        print(f"Raw product underflowed to exactly 0.0 at time step t={underflow_t} "
              f"(out of T={T}); log-space computation remained valid throughout.")
    else:
        print("Raw product did not underflow to exactly 0 for this T (try a larger T).")
    return raw_prob, log_p, underflow_t

print("\n SQ7: Raw product vs. log-space computation ")
sq7_underflow_demo(T=5000, seed=42)
```

## Simulation Question 11. 
Using the trained HMAD model from the previous simulation question, implement the full prediction pipeline and evaluate it on the held-out test set.

```python
# SQ11: Compare HMAD vs OC-SVM, required plots
print("\n SQ11: Compare HMAD vs OC-SVM Baseline")
results_oc = []
results_hmad = []
test_data_for_plots = {}

for r_test in test_anomaly_fractions:
    n_anom_test = int(100 * r_test / (1 - r_test))
    if n_anom_test == 0:
        n_anom_test = 1
    X_norm_test, X_anom_test = generate_synthetic_dataset(
        100, n_anom_test, T, anomaly_block_len, mu_anomaly, seed=TEST_SEED)
    X_test = X_norm_test + X_anom_test
    y_test = [0]*100 + [1]*n_anom_test

    # OC-SVM baseline (flattened features)
    X_train_flatten = prepare_features(X_train, mode='flatten')
    X_test_flatten = prepare_features(X_test, mode='flatten')
    auc_oc, scores_oc, _, _ = evaluate_oc_svm(X_train_flatten, X_test_flatten, y_test, nu=0.1)
    results_oc.append(auc_oc)

    # HMAD
    scores_hmad = predict_hmad(X_test, hmad_model, hmad_scaler, A_final, mus_final, sigmas_final, K=2)
    auc_hmad = roc_auc_score(y_test, scores_hmad)
    results_hmad.append(auc_hmad)

    print(f"r_test = {r_test:.2f} | OC-SVM AUC: {auc_oc:.4f} | HMAD AUC: {auc_hmad:.4f}")

    pred_oc = (scores_oc > 0).astype(int)
    pred_hmad = (scores_hmad > 0).astype(int)
    print(f"  OC-SVM  - Precision: {precision_score(y_test, pred_oc):.3f}, "
          f"Recall: {recall_score(y_test, pred_oc):.3f}, "
          f"F1: {f1_score(y_test, pred_oc):.3f}")
    print(f"  HMAD    - Precision: {precision_score(y_test, pred_hmad):.3f}, "
          f"Recall: {recall_score(y_test, pred_hmad):.3f}, "
          f"F1: {f1_score(y_test, pred_hmad):.3f}")

    
    dummy_features = []
    for seq in X_test:
        z_dummy = np.zeros(T, dtype=int)
        feat = compute_joint_features(seq, z_dummy, K=2)
        dummy_features.append(feat)
    dummy_features = np.array(dummy_features)
    dummy_features_scaled = hmad_scaler.transform(dummy_features)
    scores_dummy = hmad_model.decision_function(dummy_features_scaled)
    auc_dummy = roc_auc_score(y_test, -scores_dummy)
    print(f"  Trivial assignment AUC: {auc_dummy:.4f} (confirms Viterbi matters)")

    if r_test == 0.1:
        test_data_for_plots['X_test'] = X_test
        test_data_for_plots['y_test'] = y_test
        test_data_for_plots['scores_oc'] = scores_oc
        test_data_for_plots['scores_hmad'] = scores_hmad


if test_data_for_plots:
    y_test_01 = test_data_for_plots['y_test']
    scores_oc_01 = test_data_for_plots['scores_oc']
    scores_hmad_01 = test_data_for_plots['scores_hmad']

    fpr_oc, tpr_oc, _ = roc_curve(y_test_01, scores_oc_01)
    fpr_hm, tpr_hm, _ = roc_curve(y_test_01, scores_hmad_01)
    plt.figure()
    plt.plot(fpr_oc, tpr_oc, label=f'OC-SVM (AUC={roc_auc_score(y_test_01, scores_oc_01):.3f})')
    plt.plot(fpr_hm, tpr_hm, label=f'HMAD (AUC={roc_auc_score(y_test_01, scores_hmad_01):.3f})')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curves (r_test = 0.1)')
    plt.legend()
    plt.grid(True)
    plt.show()

    X_test = test_data_for_plots['X_test']
    y_test = test_data_for_plots['y_test']
    normal_idx = np.where(np.array(y_test) == 0)[0][0]
    anomaly_idx = np.where(np.array(y_test) == 1)[0][0]
    seq_norm = X_test[normal_idx]
    seq_anom = X_test[anomaly_idx]
    z_norm = viterbi_log(seq_norm, np.ones(2)/2, A_final, mus_final, sigmas_final)
    z_anom = viterbi_log(seq_anom, np.ones(2)/2, A_final, mus_final, sigmas_final)

    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.plot(seq_norm, 'b-', alpha=0.7, label='Observation')
    plt.plot(z_norm, 'r--', label='Viterbi state')
    plt.title('Normal Sequence')
    plt.xlabel('Time')
    plt.legend()
    plt.subplot(1, 2, 2)
    plt.plot(seq_anom, 'b-', alpha=0.7, label='Observation')
    plt.plot(z_anom, 'r--', label='Viterbi state')
    plt.title('Anomalous Sequence (block of 4.0)')
    plt.xlabel('Time')
    plt.legend()
    plt.tight_layout()
    plt.show()


plt.figure()
plt.plot(test_anomaly_fractions, results_oc, 'o-', label='OC-SVM')
plt.plot(test_anomaly_fractions, results_hmad, 's-', label='HMAD')
plt.xlabel('Test Anomaly Fraction')
plt.ylabel('ROC-AUC')
plt.title('Sensitivity to Contamination')
plt.legend()
plt.grid(True)
plt.show()
```
