## loading libraries, Dataset and preprocessing  and global variables

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score, roc_curve, precision_score, recall_score, f1_score
import warnings
warnings.filterwarnings('ignore')
import time

T = 20        
STEP = 100               
TRAIN_SEED = 1405        
TEST_SEED  = 2026
SQ6_SEED   = 7

test_anomaly_fractions = [0.01, 0.05, 0.1, 0.2]
nus = [0.01, 0.05, 0.1, 0.2,0.3,0.4]
FEATURE_COL = 'Total_Fwd_Packets'   
```

```python
# DATA LOADING & WINDOWING Continual Wind Energy (wind_random_anomalies)

df_train = pd.read_csv('datasetspart1/train.csv')
df_test  = pd.read_csv('datasetspart1/test.csv')


label_col = None
for candidate in ['label', 'anomaly', 'is_anomaly', 'Label']:
    if candidate in df_train.columns:
        label_col = candidate
        break
if label_col is None:
    for col in df_train.columns:
        if 'label' in col.lower() or 'anomaly' in col.lower():
            label_col = col
            break
if label_col is None:
    raise KeyError("Could not find a label column. Check column names: " + str(df_train.columns.tolist()))

# (0 = normal, 1 = anomaly)
df_train['binary_label'] = df_train[label_col].astype(int)
df_test['binary_label']  = df_test[label_col].astype(int)

numeric_cols = df_train.select_dtypes(include=[np.number]).columns.tolist()
for c in [label_col, 'binary_label']:
    if c in numeric_cols:
        numeric_cols.remove(c)
FEATURE_COL = numeric_cols[0]  


def create_windows_with_mask(df, feature_col):
    values = df[feature_col].values
    labels = df['binary_label'].values
    n = len(values)
    windows = []
    win_labels = []
    anomaly_masks = []                         
    for i in range(0, n - T + 1, T):
        win = values[i:i+T]
        lbl = 1 if np.any(labels[i:i+T] == 1) else 0
        windows.append(win)
        win_labels.append(lbl)
        anomaly_masks.append(labels[i:i+T] == 1)
    return np.array(windows), np.array(win_labels), anomaly_masks

df_train_normal = df_train[df_train['binary_label'] == 0]
windows_train, labels_train, _ = create_windows_with_mask(df_train_normal, FEATURE_COL)

np.random.seed(TRAIN_SEED)
n_train = min(200, len(windows_train))
train_idx = np.random.choice(len(windows_train), size=n_train, replace=False)
X_train = [windows_train[i] for i in train_idx]

windows_test, labels_test, masks_test = create_windows_with_mask(df_test, FEATURE_COL)
X_test_pool = windows_test
y_test_pool = labels_test

normal_test_idx = np.where(y_test_pool == 0)[0]
anom_test_idx   = np.where(y_test_pool == 1)[0]

print(f"Train normal windows: {len(X_train)}")
print(f"Test pool: {len(X_test_pool)} total, {y_test_pool.sum()} anomalies (frac {y_test_pool.mean():.3f})")
```

## functions

```python
# Generate a sequence from a 2-state Gaussian HMM (SQ5)
def generate_hmm_sequence(pi, A, mus, sigmas, T, seed=None):
    rng = np.random.RandomState(seed)
    K = len(mus)
    z = np.zeros(T, dtype=int)
    x = np.zeros(T)
    z[0] = rng.choice(K, p=pi)
    x[0] = rng.normal(mus[z[0]], np.sqrt(sigmas[z[0]]))
    for t in range(1, T):
        z[t] = rng.choice(K, p=A[z[t-1]])
        x[t] = rng.normal(mus[z[t]], np.sqrt(sigmas[z[t]]))
    return z, x
    
#Generate synthetic Gaussian sequences
def generate_synthetic_dataset(n_nominal, n_anomaly, T, anomaly_block_len, mu_anomaly=4.0, seed=1405):
    rng = np.random.RandomState(seed)
    X_nominal = [rng.normal(0, 1, T) for _ in range(n_nominal)]
    X_anomaly = []
    for _ in range(n_anomaly):
        seq = rng.normal(0, 1, T)
        start = rng.randint(0, T - anomaly_block_len)
        seq[start:start+anomaly_block_len] = rng.normal(mu_anomaly, 1, anomaly_block_len)
        X_anomaly.append(seq)
    return X_nominal, X_anomaly

# Log-probability of a state sequence (SQ4)
def log_prob_markov(z, pi, A):
    log_p = np.log(pi[z[0]] + 1e-15)
    for t in range(1, len(z)):
        log_p += np.log(A[z[t-1], z[t]] + 1e-15)
    return log_p

# Log-space Viterbi decoding for Gaussian HMM (SQ6)
def viterbi_log(x, pi, A, mus, sigmas):
    T_len = len(x)
    K = len(mus)
    log_emission = np.zeros((T_len, K))
    for k in range(K):
        log_emission[:, k] = -0.5 * np.log(2 * np.pi * sigmas[k]) - 0.5 * ((x - mus[k])**2 / sigmas[k])
    delta = np.zeros((T_len, K))
    psi = np.zeros((T_len, K), dtype=int)
    delta[0, :] = np.log(pi + 1e-15) + log_emission[0, :]
    for t in range(1, T_len):
        for j in range(K):
            temp = delta[t-1, :] + np.log(A[:, j] + 1e-15)
            best = np.argmax(temp)
            delta[t, j] = temp[best] + log_emission[t, j]
            psi[t, j] = best
    z_star = np.zeros(T_len, dtype=int)
    z_star[-1] = np.argmax(delta[-1, :])
    for t in range(T_len-2, -1, -1):
        z_star[t] = psi[t+1, z_star[t+1]]
    return z_star

# Normalized joint feature map Psi(x,z) (SQ9)
def compute_joint_features(x, z, K):
    T_len = len(x)
    c = np.zeros((K, K))
    for t in range(1, T_len):
        c[z[t-1], z[t]] += 1
    c_flat = c.flatten() / (T_len - 1)
    e = np.zeros(K)
    counts = np.zeros(K)
    for t in range(T_len):
        e[z[t]] += x[t]
        counts[z[t]] += 1
    e = e / (counts + 1e-15)
    return np.concatenate([c_flat, e])

# Convert list of sequences to fixed-length feature matrix
def prepare_features(sequences, mode='flatten', window_size=10):
    X = []
    for seq in sequences:
        if mode == 'flatten':
            X.append(seq)
        elif mode == 'summary':
            X.append([np.mean(seq), np.std(seq), np.min(seq), np.max(seq), np.median(seq)])
        elif mode == 'window':
            feats = []
            for w in range(0, len(seq), window_size):
                window = seq[w:w+window_size]
                feats.append(np.mean(window))
                feats.append(np.std(window))
            X.append(feats)
    return np.array(X)

# Train and evaluate OC-SVM(AUC and anomaly scores)
def evaluate_oc_svm(X_train, X_test, y_test, nu=0.1, gamma='scale', kernel='rbf'):
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    model = OneClassSVM(kernel=kernel, gamma=gamma, nu=nu)
    model.fit(X_train_scaled)
    scores = model.decision_function(X_test_scaled)
    anomaly_scores = -scores
    auc = roc_auc_score(y_test, anomaly_scores)
    return auc, anomaly_scores, model, scaler

# HMAD training loop (SQ10)
def train_hmad(X_train, K=2, nu=0.1, max_iter=10, gamma='scale', seed=1405):
    np.random.seed(seed)
    n_seq = len(X_train)
    T_len = len(X_train[0])
    pooled = np.concatenate(X_train).reshape(-1, 1)
    kmeans = KMeans(n_clusters=K, random_state=seed, n_init=10)
    kmeans.fit(pooled)
    mus = kmeans.cluster_centers_.flatten()
    idx = np.argsort(mus)
    mus = mus[idx]
    sigmas = np.zeros(K)
    labels_km = kmeans.predict(pooled)
    for k in range(K):
        cluster_data = pooled[labels_km == idx[k]]
        sigmas[k] = np.var(cluster_data) if len(cluster_data) > 1 else 1.0
    A = np.full((K, K), 0.1 / (K-1)) if K > 1 else np.array([[1.0]])
    for k in range(K):
        A[k, k] = 0.9
    pi = np.ones(K) / K
    assignment_history = []
    score_history = []
    frac_changed_history = []
    SIGMA_FLOOR = 1e-6
    for m in range(max_iter):
        z_assignments = [viterbi_log(seq, pi, A, mus, sigmas) for seq in X_train]
        features = np.array([compute_joint_features(seq, z, K) for seq, z in zip(X_train, z_assignments)])
        scaler = StandardScaler()
        features_scaled = scaler.fit_transform(features)
        model = OneClassSVM(kernel='rbf', gamma=gamma, nu=nu)
        model.fit(features_scaled)
        scores = model.decision_function(features_scaled)
        score_history.append(np.mean(scores))
        if m > 0:
            changes = sum(not np.array_equal(z_assignments[i], assignment_history[-1][i]) for i in range(n_seq))
            frac_changed = changes / n_seq
        else:
            frac_changed = 1.0
        frac_changed_history.append(frac_changed)
        assignment_history.append(z_assignments)
        if m > 0 and frac_changed < 0.01:
            print(f"HMAD converged at iteration {m+1}")
            break
        A_new = np.zeros((K, K))
        for z in z_assignments:
            for t in range(1, T_len):
                A_new[z[t-1], z[t]] += 1
        for i in range(K):
            row_sum = A_new[i, :].sum()
            A_new[i, :] = A_new[i, :] / row_sum if row_sum > 0 else 1.0/K
        A = A_new
        counts = np.zeros(K)
        sums = np.zeros(K)
        sumsq = np.zeros(K)
        for seq, z in zip(X_train, z_assignments):
            for t in range(T_len):
                k = z[t]
                counts[k] += 1
                sums[k] += seq[t]
                sumsq[k] += seq[t]**2
        for k in range(K):
            if counts[k] > 1:
                mus[k] = sums[k] / counts[k]
                sigmas[k] = max((sumsq[k] / counts[k]) - mus[k]**2, SIGMA_FLOOR)
            else:
                sigmas[k] = 1.0
    return model, scaler, A, mus, sigmas, assignment_history, score_history, frac_changed_history

# prediction pipeline for HMAD (SQ11)
def predict_hmad(X_test, model, scaler, A, mus, sigmas, K=2):
    features = []
    for seq in X_test:
        z_seq = viterbi_log(seq, np.ones(K)/K, A, mus, sigmas)
        features.append(compute_joint_features(seq, z_seq, K))
    features = np.array(features)
    features_scaled = scaler.transform(features)
    scores = model.decision_function(features_scaled)
    return -scores   # anomaly scores

# raw product underflow vs log-space (SQ7)
def sq7_underflow_demo(T_len=5000, seed=42):
    pi = np.array([0.5, 0.5])
    A = np.array([[0.9, 0.1], [0.1, 0.9]])
    rng = np.random.RandomState(seed)
    z = np.zeros(T_len, dtype=int)
    z[0] = rng.choice(2, p=pi)
    for t in range(1, T_len):
        z[t] = rng.choice(2, p=A[z[t-1]])
    raw_prob = pi[z[0]]
    underflow_t = None
    for t in range(1, T_len):
        raw_prob *= A[z[t-1], z[t]]
        if raw_prob == 0.0 and underflow_t is None:
            underflow_t = t
    log_p = log_prob_markov(z, pi, A)
    print(f"Raw product P(z) after T={T_len} steps: {raw_prob}")
    print(f"Log-space log P(z): {log_p:.4f}")
    if underflow_t is not None:
        print(f"Underflow at t={underflow_t}")
    return raw_prob, log_p, underflow_t
```

با جایگذاری در شرط مجموع:

```python
# RUNTIME
T = 20
TRAIN_SEED = 1405
K = 2

try:
    df_train = pd.read_csv('datasetspart1/train.csv')
    df_test  = pd.read_csv('datasetspart1/test.csv')

    label_col = None
    for candidate in ['label', 'anomaly', 'is_anomaly', 'Label']:
        if candidate in df_train.columns:
            label_col = candidate
            break
    if label_col is None:
        for col in df_train.columns:
            if 'label' in col.lower() or 'anomaly' in col.lower():
                label_col = col
                break
    if label_col is None:
        raise KeyError("Could not find a label column.")

    df_train['binary_label'] = df_train[label_col].astype(int)
    df_test['binary_label']  = df_test[label_col].astype(int)

    numeric_cols = df_train.select_dtypes(include=[np.number]).columns.tolist()
    for c in [label_col, 'binary_label']:
        if c in numeric_cols:
            numeric_cols.remove(c)
    FEATURE_COL = numeric_cols[0]


    def create_windows_with_mask(df, feature_col):
        values = df[feature_col].values
        labels = df['binary_label'].values
        n = len(values)
        windows = []
        win_labels = []
        for i in range(0, n - T + 1, T):
            win = values[i:i+T]
            lbl = 1 if np.any(labels[i:i+T] == 1) else 0
            windows.append(win)
            win_labels.append(lbl)
        return np.array(windows), np.array(win_labels), None


    df_train_normal = df_train[df_train['binary_label'] == 0]
    windows_train, _, _ = create_windows_with_mask(df_train_normal, FEATURE_COL)

    np.random.seed(TRAIN_SEED)
    n_train = min(200, len(windows_train))
    train_idx = np.random.choice(len(windows_train), size=n_train, replace=False)
    X_train = [windows_train[i] for i in train_idx]


except FileNotFoundError:
    rng = np.random.RandomState(TRAIN_SEED)
    t = np.arange(T)

    X_train = np.array([np.sin(2 * np.pi * t / 5) + rng.normal(0, 0.2, T) for _ in range(200)])


def prepare_features(seq_list, mode='flatten', window_size=4):
    X = []
    for seq in seq_list:
        if mode == 'flatten':          
            X.append(seq)
        elif mode == 'summary':          
            X.append([np.mean(seq), np.std(seq), np.min(seq), np.max(seq), np.median(seq)])
        elif mode == 'window':          
            feats = []
            for w in range(0, len(seq), window_size):
                window = seq[w:w+window_size]
                feats.append(np.mean(window))
                feats.append(np.std(window))
            X.append(feats)
    return np.array(X)

def viterbi_log(x, pi, A, mus, sigmas):
    T_len = len(x)
    K = len(mus)
    log_emission = np.zeros((T_len, K))
    for k in range(K):
        log_emission[:, k] = -0.5 * np.log(2 * np.pi * sigmas[k]) - 0.5 * ((x - mus[k])**2 / sigmas[k])
    delta = np.zeros((T_len, K))
    psi = np.zeros((T_len, K), dtype=int)
    delta[0, :] = np.log(pi + 1e-15) + log_emission[0, :]
    for t in range(1, T_len):
        for j in range(K):
            temp = delta[t-1, :] + np.log(A[:, j] + 1e-15)
            best = np.argmax(temp)
            delta[t, j] = temp[best] + log_emission[t, j]
            psi[t, j] = best
    z_star = np.zeros(T_len, dtype=int)
    z_star[-1] = np.argmax(delta[-1, :])
    for t in range(T_len-2, -1, -1):
        z_star[t] = psi[t+1, z_star[t+1]]
    return z_star

def compute_joint_features(x, z, K):
    T_len = len(x)
    c = np.zeros((K, K))
    for t in range(1, T_len):
        c[z[t-1], z[t]] += 1
    c_flat = c.flatten() / (T_len - 1)
    e = np.zeros(K)
    counts = np.zeros(K)
    for t in range(T_len):
        e[z[t]] += x[t]
        counts[z[t]] += 1
    e = e / (counts + 1e-15)
    return np.concatenate([c_flat, e])

def train_hmad(X_train, K=2, max_iter=10, seed=1405):
    np.random.seed(seed)
    n_seq = len(X_train)
    T_len = len(X_train[0])
    pooled = np.concatenate(X_train).reshape(-1, 1)
    kmeans = KMeans(n_clusters=K, random_state=seed, n_init=10)
    kmeans.fit(pooled)
    mus = kmeans.cluster_centers_.flatten()
    idx = np.argsort(mus)
    mus = mus[idx]
    sigmas = np.zeros(K)
    labels_km = kmeans.predict(pooled)
    for k in range(K):
        sigmas[k] = np.var(pooled[labels_km == idx[k]]) if np.any(labels_km == idx[k]) else 1.0
    A = np.full((K, K), 0.1 / (K-1)) if K > 1 else np.array([[1.0]])
    np.fill_diagonal(A, 0.9)
    pi = np.ones(K) / K
    SIGMA_FLOOR = 1e-6
    
    for m in range(max_iter):
        z_assignments = [viterbi_log(seq, pi, A, mus, sigmas) for seq in X_train]
        features = np.array([compute_joint_features(seq, z, K) for seq, z in zip(X_train, z_assignments)])
        scaler = StandardScaler()
        features_scaled = scaler.fit_transform(features)
        model = OneClassSVM(kernel='rbf', gamma='scale', nu=0.1).fit(features_scaled)
        
        A_new = np.zeros((K, K))
        for z in z_assignments:
            for t in range(1, T_len):
                A_new[z[t-1], z[t]] += 1
        A_new = A_new / A_new.sum(axis=1, keepdims=True)
        A = A_new
        
        counts = np.zeros(K)
        sums = np.zeros(K)
        sumsq = np.zeros(K)
        for seq, z in zip(X_train, z_assignments):
            for t in range(T_len):
                k = z[t]
                counts[k] += 1
                sums[k] += seq[t]
                sumsq[k] += seq[t]**2
        for k in range(K):
            if counts[k] > 1:
                mus[k] = sums[k] / counts[k]
                sigmas[k] = max((sumsq[k] / counts[k]) - mus[k]**2, SIGMA_FLOOR)
            else:
                sigmas[k] = 1.0
    return model, scaler, A, mus, sigmas


window_size = max(1, T // 5)



start_flat = time.perf_counter()
X_flat = prepare_features(X_train, mode='flatten')
scaler_flat = StandardScaler().fit(X_flat)
oc_svm = OneClassSVM(kernel='rbf', gamma='scale', nu=0.1).fit(scaler_flat.transform(X_flat))
end_flat = time.perf_counter()
dur_flat = end_flat - start_flat

start_sum = time.perf_counter()
X_sum = prepare_features(X_train, mode='summary')
scaler_sum = StandardScaler().fit(X_sum)
oc_svm = OneClassSVM(kernel='rbf', gamma='scale', nu=0.1).fit(scaler_sum.transform(X_sum))
end_sum = time.perf_counter()
dur_sum = end_sum - start_sum

start_win = time.perf_counter()
X_win = prepare_features(X_train, mode='window', window_size=window_size)
scaler_win = StandardScaler().fit(X_win)
oc_svm = OneClassSVM(kernel='rbf', gamma='scale', nu=0.1).fit(scaler_win.transform(X_win))
end_win = time.perf_counter()
dur_win = end_win - start_win

print(f"1. Flattened (D={T})    : {dur_flat:.6f} seconds")
print(f"2. Global Summary (D=5) : {dur_sum:.6f} seconds")
print(f"3. Window (W=5, D=10)   : {dur_win:.6f} seconds")

start_hmad = time.perf_counter()
hmad_model, hmad_scaler, A_final, mus_final, sigmas_final = train_hmad(X_train, K=2, max_iter=10, seed=1405)
end_hmad = time.perf_counter()
dur_hmad = end_hmad - start_hmad

print(f"4. HMAD Loop (M=10)     : {dur_hmad:.6f} seconds")


print("  WIND TURBINE RUNTIME SUMMARY ")

print(f"Parameters: n={len(X_train)}, T={T}, K={K}, Iterations=10")
print(f"1. OC-SVM (Flattened)  : {dur_flat:.4f}s  [Requirement: < 2s]")
print(f"2. OC-SVM (Summary)    : {dur_sum:.4f}s  [Requirement: < 2s]")
print(f"3. OC-SVM (Window)     : {dur_win:.4f}s  [Requirement: < 2s]")
print(f"4. HMAD Loop           : {dur_hmad:.4f}s  [Requirement: < 30s]")

```

```python

print(" Section6: WIND TURBINE BONUS PIPELINE (HMAD vs OC-SVM)")


def prepare_features(seq_list, mode='flatten', W=5):
    features = []
    for x in seq_list:
        if mode == 'flatten':
            features.append(x.flatten())

        elif mode == 'summary':
            features.append([np.mean(x), np.std(x), np.min(x), np.max(x), np.median(x)])

        elif mode == 'window':
            L = len(x) // W
            w_feats = []
            for w in range(W):
                window = x[w*L : (w+1)*L]
                w_feats.extend([np.mean(window), np.std(window)])
            features.append(w_feats)
    return np.array(features)

def evaluate_oc_svm(X_train, X_test, y_test, nu=0.1):

    clf = OneClassSVM(nu=nu, kernel='rbf', gamma='scale')
    clf.fit(X_train)
    scores = -clf.decision_function(X_test)  # Anomaly_score = -normality_score
    preds = (scores > 0).astype(int)
    auc = roc_auc_score(y_test, scores)
    return auc, scores, preds

# Data Preprocessing (6.2.2, 6.3.1)

K = 2 
test_anomaly_fractions = [0.01, 0.05, 0.1, 0.2]
TEST_SEED = 1405 
W = 5  

X_train_flat = prepare_features(X_train, mode='flatten')
X_train_sum = prepare_features(X_train, mode='summary')
X_train_win = prepare_features(X_train, mode='window', W=W)

scaler_flat = StandardScaler()
scaler_sum = StandardScaler()
scaler_win = StandardScaler()

X_train_flat_scaled = scaler_flat.fit_transform(X_train_flat)
X_train_sum_scaled = scaler_sum.fit_transform(X_train_sum)
X_train_win_scaled = scaler_win.fit_transform(X_train_win)

results_oc_flat, results_oc_sum, results_oc_win, results_hmad = [], [], [], []
roc_data = {}

for r_test in test_anomaly_fractions:
    n_anom = int(100 * r_test)
    n_norm = 100 - n_anom
    n_anom = min(n_anom, len(anom_test_idx))
    n_norm = min(n_norm, len(normal_test_idx))
    
    np.random.seed(TEST_SEED)
    ch_n = np.random.choice(normal_test_idx, size=n_norm, replace=False)
    ch_a = np.random.choice(anom_test_idx, size=n_anom, replace=False) if n_anom > 0 else []
    
    X_cur = list(X_test_pool[ch_n]) + list(X_test_pool[ch_a])
    y_cur = [0]*n_norm + [1]*n_anom

    # OC-SVM (Flattened)
    X_test_flat = scaler_flat.transform(prepare_features(X_cur, mode='flatten'))
    auc_flat, scores_flat, pred_flat = evaluate_oc_svm(X_train_flat_scaled, X_test_flat, y_cur)
    results_oc_flat.append(auc_flat)

    # OC-SVM (Summary Stats)
    X_test_sum = scaler_sum.transform(prepare_features(X_cur, mode='summary'))
    auc_sum, scores_sum, pred_sum = evaluate_oc_svm(X_train_sum_scaled, X_test_sum, y_cur)
    results_oc_sum.append(auc_sum)

    # OC-SVM (Window Summary) 
    X_test_win = scaler_win.transform(prepare_features(X_cur, mode='window', W=W))
    auc_win, scores_win, pred_win = evaluate_oc_svm(X_train_win_scaled, X_test_win, y_cur)
    results_oc_win.append(auc_win)

    scores_hmad = predict_hmad(X_cur, hmad_model, scaler_hmad, A_final, mus_final, sigmas_final, K=K)
    auc_hmad = roc_auc_score(y_cur, scores_hmad)
    results_hmad.append(auc_hmad)
    pred_hmad = (scores_hmad > 0).astype(int)

    print(f"\nr_test = {r_test:.2f}")
    print(f"  OC-SVM (Flattened)  : AUC={auc_flat:.4f}, F1={f1_score(y_cur, pred_flat):.3f}, Prec={precision_score(y_cur, pred_flat):.3f}, Rec={recall_score(y_cur, pred_flat):.3f}")
    print(f"  OC-SVM (Summary)    : AUC={auc_sum:.4f}, F1={f1_score(y_cur, pred_sum):.3f}, Prec={precision_score(y_cur, pred_sum):.3f}, Rec={recall_score(y_cur, pred_sum):.3f}")
    print(f"  OC-SVM (Window W={W}): AUC={auc_win:.4f}, F1={f1_score(y_cur, pred_win):.3f}, Prec={precision_score(y_cur, pred_win):.3f}, Rec={recall_score(y_cur, pred_win):.3f}")
    print(f"  HMAD                : AUC={auc_hmad:.4f}, F1={f1_score(y_cur, pred_hmad):.3f}, Prec={precision_score(y_cur, pred_hmad):.3f}, Rec={recall_score(y_cur, pred_hmad):.3f}")

    if np.isclose(r_test, 0.1):
        roc_data = {
            'y': y_cur, 
            'flat': scores_flat, 'sum': scores_sum, 'win': scores_win, 'hmad': scores_hmad,
            'X_test': X_cur, 'local_anom_idx': n_norm, 'global_anom_idx': ch_a
        }


if roc_data:
    y_true, pred_hmad = np.array(roc_data['y']), (roc_data['hmad'] > 0).astype(int)
    norm_idx = np.where((y_true == 0) & (pred_hmad == 0))[0][0]
    
    valid_anom = np.where((y_true == 1) & (pred_hmad == 1))[0]
    anom_idx = valid_anom[0] if len(valid_anom) > 0 else roc_data['local_anom_idx']
    
    seq_norm = X_test_pool[norm_idx]
    seq_anom = roc_data['X_test'][anom_idx]
    
    global_idx = roc_data['global_anom_idx'][anom_idx - roc_data['local_anom_idx']]
    mask_anom = masks_test[global_idx]
    change_points = np.where(np.diff(np.concatenate(([0], mask_anom.astype(int), [0]))))[0]

    fig_a, axes_a = plt.subplots(1, 2, figsize=(14, 4))
    axes_a[0].plot(seq_norm, color='navy')
    axes_a[0].set_title('Nominal Wind Turbine Sequence')
    axes_a[0].set_ylabel('Sensor Observation')
    axes_a[0].grid(alpha=0.3)
    
    axes_a[1].plot(seq_anom, color='navy')
    for i in range(0, len(change_points), 2):
        axes_a[1].axvspan(change_points[i], change_points[i+1], alpha=0.25, color='#FF6B6B', label='True Anomaly Block')
    axes_a[1].set_title('Anomalous Wind Turbine Sequence')
    axes_a[1].set_ylabel('Sensor Observation')
    axes_a[1].legend(loc='upper right')
    axes_a[1].grid(alpha=0.3)
    fig_a.suptitle('a) Sequence Visualization (r_test=0.1)', fontsize=14)
    plt.tight_layout()
    plt.show()

    # ROC Curves 
    plt.figure(figsize=(8, 6))
    fpr_f, tpr_f, _ = roc_curve(roc_data['y'], roc_data['flat'])
    fpr_s, tpr_s, _ = roc_curve(roc_data['y'], roc_data['sum'])
    fpr_w, tpr_w, _ = roc_curve(roc_data['y'], roc_data['win'])
    fpr_h, tpr_h, _ = roc_curve(roc_data['y'], roc_data['hmad'])
    
    auc_f = roc_auc_score(roc_data['y'], roc_data['flat'])
    auc_s = roc_auc_score(roc_data['y'], roc_data['sum'])
    auc_w = roc_auc_score(roc_data['y'], roc_data['win'])
    auc_h = roc_auc_score(roc_data['y'], roc_data['hmad'])
    
    plt.plot(fpr_f, tpr_f, '--', label=f'OC-SVM (Flattened), AUC={auc_f:.3f}')
    plt.plot(fpr_s, tpr_s, ':', label=f'OC-SVM (Summary), AUC={auc_s:.3f}')
    plt.plot(fpr_w, tpr_w, '-.', label=f'OC-SVM (Window W={W}), AUC={auc_w:.3f}')
    plt.plot(fpr_h, tpr_h, label=f'Simplified HMAD, AUC={auc_h:.3f}')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('b) ROC Curves (r_test=0.1) - All Feature Sets')
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

    pi = np.ones(K) / K
    z_norm = viterbi_log(seq_norm, pi, A_final, mus_final, sigmas_final)
    z_anom = viterbi_log(seq_anom, pi, A_final, mus_final, sigmas_final)


# Contamination Sensitivity Profile (All Representations)
plt.figure(figsize=(9, 5))
plt.plot(test_anomaly_fractions, results_oc_flat, 'o--', label='OC-SVM (Flattened)', lw=2)
plt.plot(test_anomaly_fractions, results_oc_sum, 's:', label='OC-SVM (Summary)', lw=2)
plt.plot(test_anomaly_fractions, results_oc_win, 'D-.', label=f'OC-SVM (Window W={W})', lw=2)
plt.plot(test_anomaly_fractions, results_hmad, '^-', label='HMAD (Wind Turbine)', lw=2)
plt.xlabel('Test Anomaly Fraction (\(r_{test}\))')
plt.ylabel('Test ROC-AUC')
plt.title('c) Wind Turbine Contamination Sensitivity Profile')
plt.legend()
plt.grid(alpha=0.4)
plt.show()
```
